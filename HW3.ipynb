```python
class Driver:

  def __init__(self, DriverID, DriverName, DriverNumber):
    self.DriverID = DriverID
    self.DriverName = DriverName
    self.DriverNumber = DriverNumber

  def ShowInfo(self):
    return f"ФИО: {self.DriverName}, ID: {self.DriverID}, номер на рации: {self.DriverNumber}"

Nobody = Driver(DriverID=0, DriverName="no", DriverNumber=0)


class Car:
  def __init__(self, Coordinates, WhoDriver, Firm, Model, Condition):
    self.Coordinates = Coordinates
    self.WhoDriver = WhoDriver
    self.Firm = Firm
    self.Model = Model
    self.Condition = Condition
  def Showinfo(self):
    if self.Condition == "в ремонте":
      return f"{self.Firm} {self.Model} в ремонте"
    else:
      if self.WhoDriver.DriverID == 0:
        return f"{self.Firm} {self.Model} на стоянке"
      else:
        return f"{self.Firm} {self.Model}; координаты: ({self.Coordinates[0]}, {self.Coordinates[1]}); ID водителя: {self.WhoDriver.DriverID}"

  def AssignDriver(self, NewDriver):
    if self.Condition == "в ремонте":
      return f"{self.Firm} {self.Model} в ремонте, водителя назначить нельзя"
    else:
      self.WhoDriver = NewDriver
      return f"Водитель с ID: {self.WhoDriver.DriverID} приступил к использованию машины {self.Firm} {self.Model}"

  def DropDriver(self):
    if self.WhoDriver.DriverID == 0:
      return f"{self.Firm} {self.Model} уже на стоянке"
    else:
      self.WhoDriver = Nobody
      return f"{self.Firm} {self.Model} сдана на стоянку"

  def StartRepair(self):
    if self.Condition == "в ремонте":
      return f"{self.Firm} {self.Model} уже в ремонте"
    else:
      self.Condition = "в ремонте"
      self.WhoDriver = Nobody
      return f"{self.Firm} {self.Model} отправлена в ремонт"

  def EndRepair(self):
    if self.Condition != "в ремонте":
      return f"{self.Firm} {self.Model} уже не находится в ремонте"
    else:
      self.Condition = "в эксплуатации"
      return f"{self.Firm} {self.Model} возвращена в эксплуатацию"


class Truck(Car):

  def __init__(self, Coordinates, WhoDriver, Firm, Model, Condition, Transporting):
    super().__init__(Coordinates, WhoDriver, Firm, Model, Condition)
    self.Transporting = Transporting

  def Showinfo(self):
    if self.Condition == "в ремонте":
      return f"{self.Firm} {self.Model} в ремонте"
    else:
      if self.WhoDriver.DriverID == 0:
        return f"{self.Firm} {self.Model} на стоянке"
      else:
        if self.Transporting == "ничего":
          return f"{self.Firm} {self.Model} ничего не перевозит; координаты: ({self.Coordinates[0]}, {self.Coordinates[1]}); ID водителя: {self.WhoDriver.DriverID}"
        else:
          return f"{self.Firm} {self.Model} перевозит {self.Transporting}; координаты: ({self.Coordinates[0]}, {self.Coordinates[1]}); ID водителя: {self.WhoDriver.DriverID}"

  def ChangeTran(self, What):
    self.Transporting = What
    if What == "ничего":
      return f"{self.Firm} {self.Model} теперь ничего не перевозит"
    else:
      return f"{self.Firm} {self.Model} теперь перевозит {self.Transporting}"


class QuarryTruck(Car):

  def __init__(self, Coordinates, WhoDriver, Firm, Model, Condition, LiftingCapacity):
    super().__init__(Coordinates, WhoDriver, Firm, Model, Condition)
    self.LiftingCapacity = LiftingCapacity

  def Showinfo(self):
    if self.Condition == "в ремонте":
      return f"{self.Firm} {self.Model} в ремонте"
    else:
      if self.WhoDriver.DriverID == 0:
        return f"{self.Firm} {self.Model} на стоянке"
      else:
        return f"{self.Firm} {self.Model} {self.LiftingCapacity}; координаты: ({self.Coordinates[0]}, {self.Coordinates[1]}); ID водителя: {self.WhoDriver.DriverID}"



```
